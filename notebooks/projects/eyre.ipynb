## Eyre
Modern software is built on enormous dependency graphs.

* A single application may rely on 100-1k+ third-party packages pulled from open ecosystems like npm, PyPI, Cargo, Maven, and others
* Packages move quickly and change constantly
* Packages are one of the largest attack surfaces in modern software

Today, most dependency governance is reactive.

* Vulnerabilities often detected after packages have already entered your systems
* Organizations are forced to develop their own package intake pipelines. No two companies operate the same way
* Package intake today requires manual review and tribal knowledge

**Eyre is a policy-driven enforcement layer for software package ecosystems.**

* Eyre automates the package intake pipeline and brings all your tools together to secure your code
* Eyre gates packages at install time, allowing for extremely powerful and expressive policies to control packages
* Our goal is to not slow developers down with ticket queues or manual security gates
* We aim to bring policy closer to engineering teams who deeply understand their codebase

**Over time, Eyre expands beyond enforcement into ecosystem intelligence, operational visibility, policy coordination, and software supply chain governance tooling.**

## How we work

### One-time Package Ingestion

When a package is needed, anyone can request it to be pulled into your Eyre instance. Eyre will automatcally trigger all  your package analysis tools and even do some basic analysis by itself.

Eyre will then save the results of your analysis tools to be used at policy enforcement time.

### Policy Enforcement

When a package is downloaded using npm, pip or another package manager, the download will be proxied through our open source package gateway.

The gateway will provide the saved analysis results and your request information to rapidly decide whether you are allowed to download a package. The gateway will also record telemetry for package usage analytics

This all happens in realtime, with minimal performance impact. Our VMs spin up in milliseconds, and we try to keep most policy decisions under 100ms.

### Recursive Policy

Today, package policy is generally *yes* or *no*. A company might have multiple products:

1. External facing website with constant user traffic and low risk tolerance
2. Internal tool, inaccessible from the internet

For #1, you might care more about XSS/RCE/SQLi vulnerabilities and those need to be remediated *immediately*. But for #2, these provide little-to-no risk. For #2, you just care whether the package contains malware or not.

Our policy engine supports nested policies: have 1+ company-wide policies, and your different orgs and apps can have their own policies.

This allows right-sized policies where they are necessary.

### Bring Cybersecurity Closer to Engineering

In rock climbing, it's a common misconception that the rope and gear is keeping the climber alive. It's not. Instead, it's the climber's technique and risk management which is the **first line of defense**. The rope exists to catch them when they fail.

We aim to bring cybersecurity and app supply chain closer to engineers. We strongly believe that the engineers who own an application are the **first line of defense**. They should have an intimate understanding of their application's dependency graph, and have a strong say in the policy applied to their app.
