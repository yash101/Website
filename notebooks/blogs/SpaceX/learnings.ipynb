Lessons learned working at SpaceX

I joined SpaceX expecting insane engineering, fighting with the bounary of physics, smart people and urgency. All those were true, but they aren't the full story of why SpaceX is successful.

This isn't an attempt to explain *the secret to SpaceX*. I was there for only a short period of time, and primarily worked on internal software and manufacturing systems. But the experience changed how I think about engineering organizations, product development, and execution.

## Reduce Organizational & Technical Latency

The meta-goal in reducing organizational and technical latency is learning how to amplify the signal and delete the noise. Your goal should be to minimize and optimize your critical path so it can be completed in a shorter period of time. And your critical path work should have JUST enough in it to investigate the next steps for your product.

### Ownership

If there's one thing to take out of this article, it's to...

> *Own something, no matter how small or big it is. Own it early*

Ownership early on is important because it rebrands you from someone closing tickets to the one who is *driving* the product and *responsible* for its outcome. Owning outcomes is how you build trust within your team, your organization, and evenutally, your company.

You know you have succeeded at owning something when:
* You are the PoC for that something
* You are DM'ed consistently for that something (alias of #1, but very strong signal)
* When someone points to that something, they mention "ask {you} for help"

### Speed is a Feature

Getting a mediocre or even a non-functioning solution delivered this week will help you understand and break assumptions.

* A 90% correct solution that people trust beats a 100% correct solution no one adopts
* An incomplete solution will provide you the useful insight on what to build next

Speed also does not have to come from just shrinking the critical path. Deleting organizational inefficiencies is another lever you can pull.

Consider:

* How do you gather information from your team and stakeholders? IRL? Video call? E-mail? Carrier pigeon?
* How do you store information so you can efficiently revisit it in the future?

In fact, my struggles with this while building Eyre led to me building my [own ERP system](https://github.com/yash101/microerp)!

### Delete Requirements

Delete requirements until it hurts. This is your critical path. Then add 1-2 back in, and those can be the icing on the cake.

Why?

Requirements are *guesses* about reality. You don't know what reality is until you ship. Thus, *you have no real requirements until you ship*. Until you ship, all *requirements* are *assumptions*.

### Delete Features and Complexity

Deleting features and complexity is an abstraction on top of deleting requirements. I separate them because it is easier to discuss them independently, but the underlying idea is the same.

Deleting features and complexity is really important. Why?

1. **Complexity compounds faster than we intuitively expect.**
    Complex applications are harder to compose together, take longer to build and slow you down to market.

2. **Every feature and component requires maintenance**
    For proof of this, try uplifting a Java Spring app from 15 years ago. Every single feature needs continuous work, uplift and testing. LLMs and AIs have somewhat reduced this problem, but it still exists.

3. **Your team has finite bandwidth**
    Your team can either allocate its effort towards building many shallow features or a few deep features.

### Story Time on Deleting

> While developing Eyre (my startup), initially I thought I needed a few features:
> 
> * CI-like workers as first-class permanent infra to ingest and analyze packages
> * Support for every ecosystem in v0 (npm, pip, docker, etc.)
> * Sandboxing multi-tenant setups using Firecracker VMs
> * Custom v8-based executor with JIT modifications for performance and other v8 modifications for further hardening
>
> I deleted all of them. Not because they were bad ideas... many may exist in the future. But **they were not on the critical path to learning whether Eyre solved a real problem.**
>
> By deleting them, I was able to focus on the core problem: realtime policy enforcement for software dependencies.
>
> Keeping them would have dramatically increased development time, operational complexity, debugging surface area, and cognitive load before the product had even met reality.
>
> Simpler systems ship faster. The faster you ship, the faster you learn and can grow.
>
> Now, Eyre works using:
> 
> * Webhooks for package intake and analysis
> * Only support for npm
> * Sandbox was deleted
> * Using node.js workers instead of C++ + v8.
>
> And, as I talk to customers with my v0 in hand, I can learn what they actually need and prioritize bringing those features to market.

### Constraints are Superpowers

When you google search "pros and cons of constraints", most articles are about how to delete constraints. There's even a [theory of constraints](https://en.wikipedia.org/wiki/Theory_of_constraints) which recommends restructuring organizations to delete constraints.

I reject that. Why?

Constraints *are* superpowers, not villains. Why? **constraints reduce search space**

Without constraints:

* Every option is possible
* Every architecture is viable
* Every framework is tempting
* Every feature seems reasonable
* Deadlines don't matter

That creates decision paralysis, complexity, fruitless local optimization and most importantly, lack of coherance.

With constraints, you gain:

* Forcd standardization
* Prioritzation
* The correct path and easy path tend to converge
* What actually matters is revealed
* Fake requirements get exposed
* Decisions get made faster

> Non-engineering example: The lack of constraints is why many often struggle to pick a restaurant, a car, or make most other decisions.

> CompSci analogy: constraints act like the heuristics in A* and help you more efficiently pathfind from start to end.

### Baseline Constraints Which Matter

* **Programming languages**: use the same language / set of languages EVERYWHERE in your stack. This allows anyone to understand and work on a codebase from day 0
* **Frameworks**: use the same framework everywhere. It doesn't matter if you use React, Angular or even jQuery + Bootstrap. Pick one and roll with it
* **Standard Libs**: build standard libraries and common paradigms to code with. Strictly enforce code guidelines and rules
* **Standards**: often, multiple standards overlap and solve the same problem. Pick one and stick with it. Example: ODB++ vs IPC-2581. Decide which one you want and just support one.
* **Vendors**: pick a vendor or two and work with them until scaling up to the point where that no longer works. Once that happens, you must define the constraints for your new manufacturers to follow.

## Question Assumptions

We **MUST** make assumptions in this to survive in this world. Otherwise, we would never accomplish anything. Why?

However, it's important to revisit these assumptions and test if they still hold today. In engineering, often those assumptions will no longer hold.

Some legendary assumptions:

* "Nobody needs more than 640KB of memory"... looks down at laptop.
* "Most people won’t buy things online"... checks Amazon

### Understand Assumptions, don't JUST Question Them

Most assumptions were made for a reason. Until you step into the shoes of the person who created the assumption, you often do not understand the original constraint, failure mode, or tradeoff that led to it.

This is especially true in software engineering, where assumptions tend to be cyclical. The industry repeatedly swings between:

* monoliths and distributed systems
* microservices and modular monoliths
* monorepos and multirepos
* thick clients and thin clients
* centralized and decentralized architectures

Why?

Because each generation solves the problems of the previous generation, then eventually creates its own problems.

For example:

Monoliths became painful as organizations scaled. Teams strugled with ownership boundaries, deployments, and coordination. Distributed systems and microservices emerged to solve those issues.

Then distributed systems introduced:

* operational complexity
* network failures
* observability challenges
* deployment orchestration
* service ownership confusion
* debugging across system boundaries

Eventually, many teams realized they had prematurely distributed their systems and began moving back toward modular monoliths.

The same thing happened with:

* monorepos vs multirepos
* synchronous vs asynchronous systems
* centralized vs federated infrastructure

Most engineering trends are not universally correct or incorrect. They are reactions to specific pain points and constraints. This is why understanding why an assumption exists is more important than blindly copying or deleting it.

## Customer / Stakeholder Feedback

Stakeholders don't know what they want. If you directly listen to customers and ask them for ideas, you'll often build features which sound cool but no one uses.

Stakeholders know their pains. Understand where they spend the most time and what gets them to want to throw their laptop out of the window. That is where you find most winning ideas and what to work on next. Often this involves pulling a chair next to your customer and watching them work.

**Book rec**: If you want to learn more about how to get customer/stakeholder feedback, I recommend reading ["The Mom Test" by Rob Fitzpatrick](https://www.momtestbook.com/).
